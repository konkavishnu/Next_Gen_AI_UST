```python
# import the necessary imports
import numpy as np
import pandas as pd
```

```python
data= pd.read_csv("LoanApprovalPrediction.csv")
data.head()
```

```python
# checking for missing values

data.isna().sum() # this func will give sum of missing values on each cols
```

NOTE : when ever we encounter missing values , we handle them by replacing those missing values by replacing the missing values with mean/median/mode of that given column is a common practice

IF DATA IS CONTINOUS :

```python
data['Dependents']=data['Dependents'].fillna(data['Dependents'].mean())
```

```python
data['LoanAmount']=data['LoanAmount'].fillna(data['LoanAmount'].mean())
```

```python
data['Loan_Amount_Term']=data['Loan_Amount_Term'].fillna(data['Loan_Amount_Term'].mean())
```

IF DATA IS NOT CONTINOUS:

```python
data['Credit_History'].unique()
```

```python
data['Credit_History']=data['Credit_History'].fillna(data['Credit_History'].mode()[0])
```

```python
data.isna().sum()
```
